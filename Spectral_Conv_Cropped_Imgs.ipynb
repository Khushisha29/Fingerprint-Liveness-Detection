# Cropping images

```python
import cv2
import numpy as np
import os

# Path to the root Testing folder
root_dir = "/kaggle/input/livdet2015/LivDet2015/Testing"
# Path to save cropped images (new root)
output_root = "/kaggle/working/Cropped_Testing"

# Function to crop fingerprint ROI using intensity sums
def crop_fingerprint(img):
    # Invert image so ridges are bright
    inv_img = cv2.bitwise_not(img)
    
    # Row-wise and column-wise sums
    row_sum = np.sum(inv_img, axis=1)
    col_sum = np.sum(inv_img, axis=0)
    
    # Thresholds (10% of max sum)
    row_thresh = np.max(row_sum) * 0.1
    col_thresh = np.max(col_sum) * 0.1
    
    rows = np.where(row_sum > row_thresh)[0]
    cols = np.where(col_sum > col_thresh)[0]
    
    if len(rows) == 0 or len(cols) == 0:
        return img  # fallback: return original if no ROI detected
    
    ymin, ymax = rows[0], rows[-1]
    xmin, xmax = cols[0], cols[-1]
    
    cropped = img[ymin:ymax, xmin:xmax]
    return cropped

# Walk through the folder recursively
for subdir, dirs, files in os.walk(root_dir):
    for file in files:
        if file.lower().endswith((".png", ".bmp")):
            file_path = os.path.join(subdir, file)
            # Load grayscale image
            img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
            if img is None:
                print(f"Failed to load {file_path}")
                continue
            
            cropped_img = crop_fingerprint(img)
            
            # Construct output path, preserving folder structure
            relative_path = os.path.relpath(subdir, root_dir)
            save_dir = os.path.join(output_root, relative_path)
            os.makedirs(save_dir, exist_ok=True)
            
            save_path = os.path.join(save_dir, file)
            cv2.imwrite(save_path, cropped_img)

print("Cropping completed. Cropped images saved under:", output_root)
```

# Cropping and resizing to 224x224, preserving aspect ratio

```python
import cv2
import numpy as np
import os

# Path to the root Testing folder
root_dir = "/kaggle/input/livdet2015/LivDet2015/Testing"
# Path to save cropped + resized images
output_root = "/kaggle/working/Cropped_resized_Testing"

# Function to crop fingerprint ROI using intensity sums
def crop_fingerprint(img):
    inv_img = cv2.bitwise_not(img)  # invert so ridges are bright
    
    row_sum = np.sum(inv_img, axis=1)
    col_sum = np.sum(inv_img, axis=0)
    
    # Thresholds (10% of max sum)
    row_thresh = np.max(row_sum) * 0.1
    col_thresh = np.max(col_sum) * 0.1
    
    rows = np.where(row_sum > row_thresh)[0]
    cols = np.where(col_sum > col_thresh)[0]
    
    if len(rows) == 0 or len(cols) == 0:
        return None  # no ROI detected
    
    ymin, ymax = rows[0], rows[-1]
    xmin, xmax = cols[0], cols[-1]
    
    cropped = img[ymin:ymax, xmin:xmax]
    return cropped

# Function to resize with aspect ratio and pad to 224x224
def resize_with_padding(img, target_size=224):
    h, w = img.shape[:2]
    scale = target_size / max(h, w)
    new_h, new_w = int(h * scale), int(w * scale)
    
    resized = cv2.resize(img, (new_w, new_h), interpolation=cv2.INTER_AREA)
    
    # create black canvas
    padded = np.zeros((target_size, target_size), dtype=img.dtype)
    
    # center placement
    y_offset = (target_size - new_h) // 2
    x_offset = (target_size - new_w) // 2
    padded[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = resized
    
    return padded

# Walk through the folder recursively
for subdir, _, files in os.walk(root_dir):
    for file in files:
        if file.lower().endswith((".png", ".bmp")):
            file_path = os.path.join(subdir, file)
            img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
            if img is None:
                print(f"Failed to load {file_path}")
                continue
            
            cropped_img = crop_fingerprint(img)
            if cropped_img is None or cropped_img.size == 0:
                print(f"Skipping empty crop: {file_path}")
                continue
            
            final_img = resize_with_padding(cropped_img, 224)
            
            # Preserve folder structure
            relative_path = os.path.relpath(subdir, root_dir)
            save_dir = os.path.join(output_root, relative_path)
            os.makedirs(save_dir, exist_ok=True)
            
            save_path = os.path.join(save_dir, file)
            cv2.imwrite(save_path, final_img)

print("✅ Cropping + resizing completed. Final images saved under:", output_root)
```

# Training on complete training set and testing on individual sensors

```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image, UnidentifiedImageError
import os, glob

class LivDetDataset(Dataset):
    def __init__(self, live_dir, fake_dir, transform=None):
        # Collect live images (bmp + png)
        self.live_images = glob.glob(os.path.join(live_dir, "*.bmp")) + \
                           glob.glob(os.path.join(live_dir, "*.png"))
        
        # Collect fake images (bmp + png, inside subfolders)
        self.fake_images = glob.glob(os.path.join(fake_dir, "*/*.bmp")) + \
                           glob.glob(os.path.join(fake_dir, "*/*.png"))
        
        self.transform = transform
        self.data = [(img, 1) for img in self.live_images] + \
                    [(img, 0) for img in self.fake_images]

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img_path, label = self.data[idx]

        try:
            image = Image.open(img_path).convert("RGB")
        except (OSError, UnidentifiedImageError) as e:
            print(f"[WARNING] Skipping corrupted file: {img_path} ({e})")
            # Instead of crashing, return a random valid sample
            return self.__getitem__((idx + 1) % len(self.data))

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label, dtype=torch.long)


# ===== Transformations =====
# transform = transforms.Compose([
#     transforms.Resize((224, 224)),
#     transforms.ToTensor()
# ])

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5])
])

```

```python
# ===== Training dataset: combine all sensors =====
train_root = "/kaggle/working/Cropped_resized_Training"
sensors = ["CrossMatch", "GreenBit", "Digital_Persona", "Hi_Scan"]

train_datasets = []
for sensor in sensors:
    live_dir = os.path.join(train_root, sensor, sensor, "Live")
    fake_dir = os.path.join(train_root, sensor, sensor, "Fake")
    train_datasets.append(LivDetDataset(live_dir, fake_dir, transform=transform))

# Concatenate all sensors into one training dataset
train_dataset = torch.utils.data.ConcatDataset(train_datasets)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

# ===== Testing datasets: one DataLoader per sensor =====
test_root = "/kaggle/working/Cropped_resized_Testing"
test_loaders = {}

for sensor in sensors:
    live_dir = os.path.join(test_root, sensor, sensor, "Live")
    fake_dir = os.path.join(test_root, sensor, sensor, "Fake")
    dataset = LivDetDataset(live_dir, fake_dir, transform=transform)
    test_loaders[sensor] = DataLoader(dataset, batch_size=32, shuffle=False)

# Example: iterate over each test set
for sensor, loader in test_loaders.items():
    print(f"{sensor} -> {len(loader.dataset)} test samples")
```

```python

# === LivDetDataset with label float32 and shape [1] for BCE ===
class LivDetDataset(Dataset):
    def __init__(self, root, transform=None):
        self.samples = []
        self.transform = transform

        # Walk through root directory to find Live and Fake folders
        for dirpath, dirnames, filenames in os.walk(root):
            if os.path.basename(dirpath) == "Live":
                for fname in filenames:
                    if fname.endswith((".bmp", ".png")):
                        self.samples.append((os.path.join(dirpath, fname), 1))
            elif os.path.basename(os.path.dirname(dirpath)) == "Fake":
                for fname in filenames:
                    if fname.endswith((".bmp", ".png")):
                        self.samples.append((os.path.join(dirpath, fname), 0))


    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        try:
            img = Image.open(img_path).convert("RGB")
        except Exception as e:
            print(f"Skipping broken image: {img_path} ({e})")
            return self.__getitem__((idx + 1) % len(self.samples))
        if self.transform:
            img = self.transform(img)
        return img, torch.tensor([label], dtype=torch.float32)  # shape: [1]

```

```python
def evaluate(model, dataloader, device, threshold=0.5, verbose=True):
    model.eval()
    correct = 0
    total = 0
    false_accepts = 0  # fake accepted as live (FP)
    false_rejects = 0  # live rejected as fake (FN)

    with torch.no_grad():
        for imgs, labels in dataloader:
            imgs = imgs.to(device)
            labels = labels.to(device).float().view(-1, 1)

            outputs = model(imgs)
            probs = torch.sigmoid(outputs)
            preds = (probs >= threshold).float()

            correct += (preds == labels).sum().item()
            total += labels.size(0)

            false_accepts += ((preds == 1) & (labels == 0)).sum().item()
            false_rejects += ((preds == 0) & (labels == 1)).sum().item()

    accuracy = correct / total
    far = false_accepts / total  # False Acceptance Rate
    frr = false_rejects / total  # False Rejection Rate
    ace = 0.5 * (far + frr)

    if verbose:
        print(f"Total samples: {total}")
        print(f"Accuracy: {accuracy:.4f}")
        print(f"FAR (Fake accepted as live): {far:.4f}")
        print(f"FRR (Live rejected as fake): {frr:.4f}")
        print(f"ACE (Average Classification Error): {ace:.4f}")

    return accuracy, ace, false_accepts, false_rejects
```

```python
class LivDetDataset(Dataset):
    def __init__(self, root, transform=None):
        self.samples = []
        self.labels = []   # <--- add this
        self.transform = transform

        # Walk through root directory to find Live and Fake folders
        for dirpath, dirnames, filenames in os.walk(root):
            if os.path.basename(dirpath) == "Live":
                for fname in filenames:
                    if fname.endswith((".bmp", ".png")):
                        self.samples.append((os.path.join(dirpath, fname), 1))
                        self.labels.append(1)   # <--- keep labels separately
            elif os.path.basename(os.path.dirname(dirpath)) == "Fake":
                for fname in filenames:
                    if fname.endswith((".bmp", ".png")):
                        self.samples.append((os.path.join(dirpath, fname), 0))
                        self.labels.append(0)   # <--- keep labels separately

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        try:
            img = Image.open(img_path).convert("RGB")  # RGB for ResNet
        except Exception as e:
            print(f"Skipping broken image: {img_path} ({e})")
            return self.__getitem__((idx + 1) % len(self.samples))

        if self.transform:
            img = self.transform(img)

        return img, torch.tensor(label, dtype=torch.float32)  # shape: []
```

```python
def evaluate(model, dataloader, device, threshold=0.5, verbose=True):
    model.eval()
    correct = 0
    total = 0
    false_accepts = 0  # fake accepted as live (FP)
    false_rejects = 0  # live rejected as fake (FN)

    with torch.no_grad():
        for imgs, labels in dataloader:
            imgs = imgs.to(device)
            labels = labels.to(device).float()  # shape [batch]

            outputs = model(imgs).squeeze()     # shape [batch]
            probs = torch.sigmoid(outputs)
            preds = (probs >= threshold).float()

            correct += (preds == labels).sum().item()
            total += labels.size(0)

            false_accepts += ((preds == 1) & (labels == 0)).sum().item()
            false_rejects += ((preds == 0) & (labels == 1)).sum().item()

    accuracy = correct / total
    far = false_accepts / total
    frr = false_rejects / total
    ace = 0.5 * (far + frr)

    if verbose:
        print(f"Total samples: {total}")
        print(f"Accuracy: {accuracy:.4f}")
        print(f"FAR (Fake accepted as live): {far:.4f}")
        print(f"FRR (Live rejected as fake): {frr:.4f}")
        print(f"ACE (Average Classification Error): {ace:.4f}")

    return accuracy, ace, false_accepts, false_rejects
```

```python
import os
import random
from torchvision.utils import save_image

# === Path to store augmented images ===
augmented_dir = "/kaggle/working/Cropped_resized_Training_Augmented"
os.makedirs(augmented_dir, exist_ok=True)

# === Augmentation transform (without ToTensor/Normalize for saving) ===
augment_only = transforms.Compose([
    transforms.RandomRotation(degrees=10),
    transforms.RandomResizedCrop(224, scale=(0.9, 1.0)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.RandomHorizontalFlip(p=0.5),
])

# Load base dataset once (grayscale images)
base_dataset = LivDetDataset(root_dir, transform=transforms.ToTensor())

# Randomly sample 500 indices
aug_indices = random.sample(range(len(base_dataset)), 500)

print(f"Generating {len(aug_indices)} augmented samples...")

for i, idx in enumerate(aug_indices):
    img, label = base_dataset[idx]
    pil_img = transforms.ToPILImage()(img)  # convert back to PIL for augmentation
    aug_img = augment_only(pil_img)         # apply augmentation

    # Save in corresponding class folder (0=fake, 1=live)
    class_folder = os.path.join(augmented_dir, str(label))
    os.makedirs(class_folder, exist_ok=True)

    save_path = os.path.join(class_folder, f"aug_{i}.png")
    save_image(transforms.ToTensor()(aug_img), save_path)

print("✅ Augmented dataset created!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, random_split

# Paths
data_dir = "/kaggle/working/Cropped_resized_Training_Augmented"  # fixed dataset with aug copies

# Transforms (only normalization now, NO augmentation here)
train_transforms = transforms.Compose([
    # transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load dataset
dataset = datasets.ImageFolder(root=data_dir, transform=train_transforms)

# Train-Validation split
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

# Model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = models.resnet18(pretrained=True)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, len(dataset.classes))
model = model.to(device)

# Loss & Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# Training Loop
num_epochs = 15
for epoch in range(num_epochs):
    model.train()
    train_loss, correct, total = 0, 0, 0
    
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    
    train_acc = 100 * correct / total

    # Validation
    model.eval()
    val_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
    
    val_acc = 100 * correct / total

    print(f"Epoch {epoch+1}/{num_epochs}, "
          f"Train Loss: {train_loss/len(train_loader):.4f}, Train Acc: {train_acc:.2f}%, "
          f"Val Loss: {val_loss/len(val_loader):.4f}, Val Acc: {val_acc:.2f}%")
```
